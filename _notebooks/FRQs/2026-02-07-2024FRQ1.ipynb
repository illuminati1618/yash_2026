## Part A: simulateOneDay Method

This method simulates one day at the bird feeder with either birds (95% chance) or a bear (5% chance) visiting.

```java
public class Feeder {
    // Instance variable
    private int currentFood;

    // Constructor
    public Feeder(int initialFood) {
        this.currentFood = initialFood;
    }

    /**
     * Simulates one day with numBirds birds or possibly a bear at the bird feeder,
     * as described in part (a)
     * Precondition: numBirds > 0
     */
    public void simulateOneDay(int numBirds) {
        // Check for bear (5% chance)
        double chance = Math.random();
        
        if (chance < 0.05) {
            // Bear empties the feeder
            currentFood = 0;
        } else {
            // Birds visit - each bird eats random amount from 10-50 grams
            int gramsPerBird = (int)(Math.random() * 41) + 10; // 10 to 50 inclusive
            int totalEaten = gramsPerBird * numBirds;
            
            // Update currentFood
            if (totalEaten > currentFood) {
                currentFood = 0; // Birds empty the feeder
            } else {
                currentFood -= totalEaten;
            }
        }
    }

    // Getter for testing
    public int getCurrentFood() {
        return currentFood;
    }
}

// Test Part A
Feeder feeder1 = new Feeder(500);
System.out.println("Initial food: " + feeder1.getCurrentFood());
feeder1.simulateOneDay(12);
System.out.println("Food after day 1: " + feeder1.getCurrentFood());
feeder1.simulateOneDay(12);
System.out.println("Food after day 2: " + feeder1.getCurrentFood());
```

## Part B: simulateManyDays Method
