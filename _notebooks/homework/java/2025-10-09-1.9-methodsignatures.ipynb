## Popcorn Hack 1: Write and test max overloads

```java
public class PopcornMax {
    // int version of max
    public static int max(int a, int b) {
        return a >= b ? a : b;
    }

    // double version of max
    public static double max(double a, double b) {
        return a >= b ? a : b;
    }

    public static void main(String[] args) {
        System.out.println(max(3, 9));      // expected: 9
        System.out.println(max(-2, -7));    // expected: -2
        System.out.println(max(3.5, 2.9));  // expected: 3.5
        System.out.println(max(2, 2.0));    // should call double version → 2.0
    }
}

PopcornMax.main(null);
```

```java
public class PopcornPrint {
    // print method for int
    static void print(int n) {
        System.out.println("int:" + n);
    }

    // print method for String
    static void print(String s) {
        System.out.println("str:" + s);
    }

    public static void main(String[] args) {
        print(42);          // expected: int:42
        print("hello");     // expected: str:hello
        print('A' + "!");   // char + String → String → expected: str:A!
    }
}

PopcornPrint.main(null);
```

## Short Answer Questions

**1. Why `int sum(int a, int b)` and `double sum(int a, int b)` cannot both exist:**

These methods have identical signatures (name + parameter types) but different return types. In Java, return type is not part of the method signature for overloading purposes, so the compiler cannot distinguish between them when called.

**2. Parameters vs Arguments:**

Parameters are the variables declared in the method definition, while arguments are the actual values passed to the method when it's called.

## Coding Task 1: Three overloads of abs

```java
public class AbsOverloads {
    // int version of abs
    public static int abs(int x) {
        return x < 0 ? -x : x;
    }
    
    // double version of abs
    public static double abs(double x) {
        return x < 0 ? -x : x;
    }
    
    // long version of abs
    public static long abs(long x) {
        return x < 0 ? -x : x;
    }
    
    public static void main(String[] args) {
        System.out.println("abs(-5) = " + abs(-5));           // int version
        System.out.println("abs(-3.14) = " + abs(-3.14));     // double version
        System.out.println("abs(-123456789L) = " + abs(-123456789L)); // long version
        
        System.out.println("abs(5) = " + abs(5));             // int version
        System.out.println("abs(3.14) = " + abs(3.14));       // double version
        System.out.println("abs(123456789L) = " + abs(123456789L));   // long version
    }
}

AbsOverloads.main(null);
```

## Coding Task 2: Two overloads of concat

```java
public class ConcatOverloads {
    // concat two strings
    public static String concat(String a, String b) {
        return a + b;
    }
    
    // concat string repeated n times
    public static String concat(String a, int n) {
        String result = "";
        for (int i = 0; i < n; i++) {
            result += a;
        }
        return result;
    }
    
    public static void main(String[] args) {
        System.out.println("concat(\"Hello\", \" World\") = " + concat("Hello", " World"));
        System.out.println("concat(\"Java\", 3) = " + concat("Java", 3));
        System.out.println("concat(\"AB\", \"CD\") = " + concat("AB", "CD"));
        System.out.println("concat(\"*\", 5) = " + concat("*", 5));
    }
}

ConcatOverloads.main(null);
```

## FRQ 1: indexOf overloads

```java
public class IndexOfOverloads {
    /**
     * Returns the index of the first occurrence of target character in string s
     * @param target the character to search for
     * @param s the string to search in
     * @return index of first occurrence or -1 if not found
     * Assumptions: s is not null
     */
    public static int indexOf(char target, String s) {
        for (int i = 0; i < s.length(); i++) {
            if (s.charAt(i) == target) {
                return i;
            }
        }
        return -1;
    }
    
    /**
     * Returns the index of the first occurrence of target substring in string s
     * @param target the substring to search for
     * @param s the string to search in  
     * @return index of first occurrence or -1 if not found
     * Assumptions: s and target are not null, target is not empty
     * Constraints: Does not use built-in indexOf method
     */
    public static int indexOf(String target, String s) {
        if (target.length() > s.length()) {
            return -1;
        }
        
        for (int i = 0; i <= s.length() - target.length(); i++) {
            boolean found = true;
            for (int j = 0; j < target.length(); j++) {
                if (s.charAt(i + j) != target.charAt(j)) {
                    found = false;
                    break;
                }
            }
            if (found) {
                return i;
            }
        }
        return -1;
    }
    
    public static void main(String[] args) {
        // Test char version
        System.out.println("indexOf('a', \"banana\") = " + indexOf('a', "banana"));  // 1
        System.out.println("indexOf('x', \"banana\") = " + indexOf('x', "banana"));  // -1
        
        // Test String version  
        System.out.println("indexOf(\"an\", \"banana\") = " + indexOf("an", "banana"));  // 1
        System.out.println("indexOf(\"xyz\", \"banana\") = " + indexOf("xyz", "banana"));  // -1
        System.out.println("indexOf(\"ana\", \"banana\") = " + indexOf("ana", "banana"));  // 1
    }
}

IndexOfOverloads.main(null);
```

## FRQ 2: Overload clamp

```java
public class ClampOverloads {
    /**
     * Clamps an integer value to the range [low, high]
     * @param value the value to clamp
     * @param low the minimum allowed value
     * @param high the maximum allowed value
     * @return value constrained to [low, high]
     */
    public static int clamp(int value, int low, int high) {
        // Handle case where low > high by swapping
        if (low > high) {
            int temp = low;
            low = high;
            high = temp;
        }
        
        if (value < low) {
            return low;
        } else if (value > high) {
            return high;
        } else {
            return value;
        }
    }
    
    /**
     * Clamps a double value to the range [low, high]
     * @param value the value to clamp
     * @param low the minimum allowed value
     * @param high the maximum allowed value
     * @return value constrained to [low, high]
     */
    public static double clamp(double value, double low, double high) {
        // Handle case where low > high by swapping
        if (low > high) {
            double temp = low;
            low = high;
            high = temp;
        }
        
        if (value < low) {
            return low;
        } else if (value > high) {
            return high;
        } else {
            return value;
        }
    }
    
    public static void main(String[] args) {
        // Test int version
        System.out.println("clamp(5, 1, 10) = " + clamp(5, 1, 10));      // 5
        System.out.println("clamp(-2, 1, 10) = " + clamp(-2, 1, 10));    // 1
        System.out.println("clamp(15, 1, 10) = " + clamp(15, 1, 10));    // 10
        System.out.println("clamp(5, 10, 1) = " + clamp(5, 10, 1));      // 5 (swapped bounds)
        
        // Test double version
        System.out.println("clamp(5.5, 1.0, 10.0) = " + clamp(5.5, 1.0, 10.0));    // 5.5
        System.out.println("clamp(-2.3, 1.0, 10.0) = " + clamp(-2.3, 1.0, 10.0));  // 1.0
        System.out.println("clamp(15.7, 1.0, 10.0) = " + clamp(15.7, 1.0, 10.0));  // 10.0
        System.out.println("clamp(5.5, 10.0, 1.0) = " + clamp(5.5, 10.0, 1.0));    // 5.5 (swapped bounds)
    }
}

ClampOverloads.main(null);
```
