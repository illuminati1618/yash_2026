## Popcorn Hack 2

```java
// Popcorn Hack #2: Complete the Book class

public class Book {
    // TODO: Add 3 attributes (title, author, pages)
    private String title;
    private String author;
    private int pages;
    
    // TODO: Add constructor
    public Book(String title, String author, int pages) {
        this.title = title;
        this.author = author;
        this.pages = pages;
    }
    
    // TODO: Add displayInfo() method
    public void displayInfo() {
        System.out.println("Title: " + title);
        System.out.println("Author: " + author);
        System.out.println("Pages: " + pages);
    }
    
    // TODO: Add isLong() method (returns true if pages > 300)
    public boolean isLong() {
        return pages > 300;
    }
}

// TODO: Create a Book object and test all methods
// Example: Book myBook = new Book("Java Basics", "John Doe", 350);

class BookTest {
    public static void main(String[] args) {
        // Create a Book object
        Book myBook = new Book("Java Basics", "John Doe", 350);
        
        // Test displayInfo()
        myBook.displayInfo();
        
        // Test isLong()
        System.out.println("Is this a long book? " + myBook.isLong());
        
        // Create another book for comparison
        System.out.println("\n--- Another Book ---");
        Book shortBook = new Book("Quick Guide", "Jane Smith", 150);
        shortBook.displayInfo();
        System.out.println("Is this a long book? " + shortBook.isLong());
    }
}

BookTest.main(null);
```

## Homework Hack

```java
// Homework Hack: Complete the Phone class

import java.util.ArrayList;

public class Phone {
    // TODO: Add 4 attributes
    private String brand;
    private String model;
    private int batteryLevel;
    private ArrayList<String> contacts;
    
    // TODO: Add constructor
    public Phone(String brand, String model) {
        this.brand = brand;
        this.model = model;
        this.batteryLevel = 100;  // Battery starts at 100%
        this.contacts = new ArrayList<>();  // Initialize empty contacts list
    }
    
    // TODO: Add displayInfo() method
    public void displayInfo() {
        System.out.println("=== Phone Information ===");
        System.out.println("Brand: " + brand);
        System.out.println("Model: " + model);
        System.out.println("Battery Level: " + batteryLevel + "%");
        System.out.println("Number of Contacts: " + contacts.size());
    }
    
    // TODO: Add addContact(String name) method
    public void addContact(String name) {
        contacts.add(name);
        System.out.println("Added contact: " + name);
    }
    
    // TODO: Add showContacts() method
    public void showContacts() {
        System.out.println("=== Contacts List ===");
        if (contacts.isEmpty()) {
            System.out.println("No contacts saved.");
        } else {
            for (int i = 0; i < contacts.size(); i++) {
                System.out.println((i + 1) + ". " + contacts.get(i));
            }
        }
    }
    
    // TODO: Add usePhone(int minutes) method
    public void usePhone(int minutes) {
        batteryLevel -= minutes;
        // Prevent battery from going below 0
        if (batteryLevel < 0) {
            batteryLevel = 0;
        }
        System.out.println("Used phone for " + minutes + " minutes. Battery now at " + batteryLevel + "%");
    }
}

// Test your Phone class
class PhoneTest {
    public static void main(String[] args) {
        // TODO: Create 2 Phone objects
        System.out.println("Creating Phone 1...");
        Phone phone1 = new Phone("Apple", "iPhone 15");
        System.out.println();
        
        System.out.println("Creating Phone 2...");
        Phone phone2 = new Phone("Samsung", "Galaxy S24");
        System.out.println();
        
        // TODO: Add 3 contacts to each
        System.out.println("--- Adding Contacts to Phone 1 ---");
        phone1.addContact("Alice Johnson");
        phone1.addContact("Bob Smith");
        phone1.addContact("Charlie Brown");
        System.out.println();
        
        System.out.println("--- Adding Contacts to Phone 2 ---");
        phone2.addContact("Diana Prince");
        phone2.addContact("Evan Davis");
        phone2.addContact("Fiona Green");
        System.out.println();
        
        // TODO: Use phones for some minutes
        System.out.println("--- Using Phones ---");
        phone1.usePhone(25);
        phone1.usePhone(15);
        System.out.println();
        
        phone2.usePhone(30);
        phone2.usePhone(20);
        System.out.println();
        
        // TODO: Display all information
        System.out.println("\n--- PHONE 1 STATUS ---");
        phone1.displayInfo();
        System.out.println();
        phone1.showContacts();
        
        System.out.println("\n--- PHONE 2 STATUS ---");
        phone2.displayInfo();
        System.out.println();
        phone2.showContacts();
    }
}

PhoneTest.main(null);
```
