## Example Class 2: Counter

A simple counter class and usage example.

```java
public class Counter {
    private int count;
    public void add(int x) {
        count += x;
    }
    public int getCount() {
        return count;
    }
}

// Use it:
Counter c = new Counter();
c.add(5);
int val = c.getCount();  // returns 5
System.out.println("Counter value: " + val);
```

## Popcorn Hack #2: Extended Counter

Add subtraction, multiplication, and division methods and call all five methods outside the class.

```java
// Work by [Your Name Here]
public class CounterExt {
    private int count;
    public void add(int x) { count += x; }
    public void subtract(int x) { count -= x; }
    public void multiply(int x) { count *= x; }
    public void divide(int x) { if (x != 0) count /= x; }
    public int getCount() { return count; }
}

// Use it:
CounterExt ce = new CounterExt();
ce.add(10);
ce.subtract(3);
ce.multiply(2);
ce.divide(7);
System.out.println("CounterExt final: " + ce.getCount());
```

## Homework Hack: Student Grade Tracker

```java
// Sample Student class skeleton
public class Student {
    String name;
    int totalPoints;
    int numAssignments;

    public Student(String name) {
        this.name = name;
        this.totalPoints = 0;
        this.numAssignments = 0;
    }

    public void addGrade(int points) {
        totalPoints += points;
        numAssignments++;
    }

    public double getAverage() {
        return numAssignments == 0 ? 0.0 : (double) totalPoints / numAssignments;
    }

    public String getLetterGrade() {
        double avg = getAverage();
        if (avg >= 90) return "A";
        if (avg >= 80) return "B";
        if (avg >= 70) return "C";
        if (avg >= 60) return "D";
        return "F";
    }

    public void printReport() {
        System.out.println("Student Name: " + name);
        System.out.println("Total Points: " + totalPoints);
        System.out.println("Assignments Completed: " + numAssignments);
        System.out.printf("Current Average: %.2f\n", getAverage());
        System.out.println("Letter Grade: " + getLetterGrade());
    }
}

// Sample usage
Student s1 = new Student("Emma Rodriguez");
s1.addGrade(95);
s1.addGrade(88);
s1.addGrade(92);
s1.addGrade(85);
s1.printReport();
```
