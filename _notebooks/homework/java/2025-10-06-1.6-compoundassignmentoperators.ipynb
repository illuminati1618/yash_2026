## Popcorn Hack 1

```java
public class PopcornHack1 {
    public static void main(String[] args) {
        int playerScore = 1000;
        int playerHealth = 100;
        int enemiesDefeated = 0;

        // Player defeats an enemy worth 250 points
        playerScore += 250;  // Changed from: playerScore = playerScore + 250

        // Player takes 15 damage
        playerHealth -= 15;  // Changed from: playerHealth = playerHealth - 15

        // Enemy count goes up
        enemiesDefeated++;   // Changed from: enemiesDefeated = enemiesDefeated + 1

        // Boss battle: double the current score!
        playerScore *= 2;    // Changed from: playerScore = playerScore * 2

        // Healing potion restores health to 80% of current
        playerHealth *= 4;
        playerHealth /= 5;   // 4/5 = 0.8, but we need integer math

        // Print final results
        System.out.println("Final Player Score: " + playerScore);
        System.out.println("Final Player Health: " + playerHealth);
        System.out.println("Enemies Defeated: " + enemiesDefeated);
    }
}

PopcornHack1.main(null);
```

## Popcorn Hack 2

```java
public class PopcornHack2 {
    public static void main(String[] args) {
        int score = 100;
        System.out.println("Starting score: " + score);
        
        // Deduct points for a wrong answer
        score -= 15;
        System.out.println("After wrong answer (-= 15): " + score);
        
        // Double the score with a power-up
        score *= 2;
        System.out.println("After power-up (*= 2): " + score);
        
        // Add bonus points for completing a level
        score += 50;
        System.out.println("After level completion (+= 50): " + score);
        
        // Find remainder after dividing by 7
        score %= 7;
        System.out.println("After modulo 7 (%= 7): " + score);
        
        System.out.println("\nFinal score: " + score);
    }
}

PopcornHack2.main(null);
```

## Homework Hack - Option 4

```java
public class FitnessProgressTracker {
    public static void main(String[] args) {
        // Initialize variables
        int totalCalories = 0;
        int workoutDays = 0;
        int stepCount = 3000;
        double currentWeight = 180.0;
        double weightGoal = 170.0;
        
        System.out.println("=== FITNESS PROGRESS TRACKER ===\n");
        System.out.println("Starting weight: " + currentWeight + " lbs");
        System.out.println("Weight goal: " + weightGoal + " lbs");
        System.out.println("Starting steps today: " + stepCount);
        System.out.println("Workout days this week: " + workoutDays + "\n");
        
        // Operation 1: Breakfast calories
        System.out.println("Breakfast consumed!");
        totalCalories += 400;  // Add breakfast calories
        System.out.println("Calories: " + totalCalories + " (+400 from breakfast)\n");
        
        // Operation 2: Lunch calories
        System.out.println("Lunch consumed!");
        totalCalories += 650;  // Add lunch calories
        System.out.println("Calories: " + totalCalories + " (+650 from lunch)\n");
        
        // Operation 3: Morning workout - burn calories
        System.out.println("Morning workout completed!");
        totalCalories -= 300;  // Burn calories from exercise
        workoutDays++;  // Increment workout days
        stepCount += 5000;  // Add steps from workout
        System.out.println("Calories: " + totalCalories + " (-300 burned)");
        System.out.println("Workout days: " + workoutDays);
        System.out.println("Steps: " + stepCount + "\n");
        
        // Operation 4: Snack time
        System.out.println("Afternoon snack...");
        totalCalories += 200;
        System.out.println("Calories: " + totalCalories + " (+200 from snack)\n");
        
        // Operation 5: Evening run - very active!
        System.out.println("Evening run - extra active!");
        totalCalories -= 450;  // Burn more calories
        stepCount *= 2;  // Doubled steps from active day
        workoutDays++;
        System.out.println("Calories: " + totalCalories + " (-450 burned)");
        System.out.println("Steps: " + stepCount + " (doubled from activity!)");
        System.out.println("Workout days: " + workoutDays + "\n");
        
        // Operation 6: Dinner
        System.out.println("Dinner consumed!");
        totalCalories += 550;
        System.out.println("Calories: " + totalCalories + " (+550 from dinner)\n");
        
        // Operation 7: Calculate daily averages
        System.out.println("Calculating weekly statistics...");
        int avgDailySteps = stepCount / 7;  // Average over a week
        System.out.println("Average daily steps (7-day): " + avgDailySteps);
        
        // Operation 8: Check calorie balance
        int calorieBalance = totalCalories - 2000;  // Daily goal is 2000
        if (calorieBalance > 0) {
            System.out.println("Calorie surplus: +" + calorieBalance + " calories");
        } else {
            System.out.println("Calorie deficit: " + calorieBalance + " calories");
        }
        
        // Operation 9: Calculate weight progress
        double weightLost = currentWeight - weightGoal;
        System.out.println("Weight to lose: " + weightLost + " lbs");
        
        // Operation 10: Weekly goal progress using modulo
        int progressDay = workoutDays % 7;  // Which day of the week
        System.out.println("Workout progress (day of week): Day " + progressDay);
        
        // Calculate projected weight change (rough estimate)
        double projectedWeightChange = (double)calorieBalance / 3500;  // 3500 cal = 1 lb
        double projectedWeight = currentWeight - projectedWeightChange;
        System.out.println("Projected weight after today: " + String.format("%.1f", projectedWeight) + " lbs");
        
        // Final summary
        System.out.println("\n=== DAILY SUMMARY ===");
        System.out.println("Total steps: " + stepCount);
        System.out.println("Net calories: " + totalCalories);
        System.out.println("Workouts completed: " + workoutDays);
        System.out.println("Current weight: " + currentWeight + " lbs");
        System.out.println("Goal weight: " + weightGoal + " lbs");
        
        if (workoutDays >= 5) {
            System.out.println("Status: On track! Keep it up! 💪");
        } else {
            System.out.println("Status: Good start! Try to workout more this week!");
        }
    }
}

FitnessProgressTracker.main(null);
```
