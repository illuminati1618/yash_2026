## Challenge 1: Document a Recursive Method

Below is a fully documented recursive Java method (binary search) including:
- clear description
- parameters and return
- example usage
- base case and recursive case explanation
- complexity analysis

```java
/*
 * BinarySearchRecursive.java
 *
 * Demonstrates a fully documented recursive binary search method.
 *
 * Description:
 * - binarySearch: searches a sorted integer array for a target value using recursion.
 *
 * Parameters:
 * - arr: sorted array of integers (ascending order)
 * - target: integer value to search for
 * - left: left index of current search range (inclusive)
 * - right: right index of current search range (inclusive)
 *
 * Returns:
 * - index of target if found (int)
 * - -1 if target is not found in the array
 *
 * Example Usage:
 * int[] data = {1,3,5,7,9};
 * int idx = binarySearch(data, 7, 0, data.length-1); // returns 3
 *
 * Base case(s):
 * - If left > right: search interval is empty => return -1 (not found).
 * - If middle element equals target: return middle index.
 *
 * Recursive case:
 * - If target < middle element: search left half by calling binarySearch with right = mid-1.
 * - If target > middle element: search right half by calling binarySearch with left = mid+1.
 *
 * Complexity Analysis:
 * - Time complexity: O(log n) in the worst-case because each recursive step halves the search range.
 * - Space complexity: O(log n) due to recursion depth (call stack), though an iterative binary search would be O(1).
 */

public class BinarySearchRecursive {

    /**
     * Searches for target in a sorted integer array using recursion.
     * @param arr sorted array of integers (ascending)
     * @param target the value to search for
     * @param left left index of current search interval (inclusive)
     * @param right right index of current search interval (inclusive)
     * @return index of target if found; -1 otherwise
     *
     * Preconditions: arr must be sorted in non-decreasing order.
     */
    public static int binarySearch(int[] arr, int target, int left, int right) {
        // Base case: empty interval => not found
        if (left > right) {
            return -1;
        }

        int mid = left + (right - left) / 2; // avoid overflow

        // Found the target
        if (arr[mid] == target) {
            return mid;
        }

        // Recurse into left half
        if (target < arr[mid]) {
            return binarySearch(arr, target, left, mid - 1);
        }

        // Recurse into right half
        return binarySearch(arr, target, mid + 1, right);
    }

    // Small helper to demonstrate usage
    public static void main(String[] args) {
        int[] data = {1, 3, 5, 7, 9, 11, 13};
        int target = 7;
        int idx = binarySearch(data, target, 0, data.length - 1);
        System.out.println("Searching for " + target + " in " + java.util.Arrays.toString(data));
        if (idx >= 0) {
            System.out.println("Found at index: " + idx + " (value=" + data[idx] + ")");
        } else {
            System.out.println("Not found");
        }
    }
}
BinarySearchRecursive.main(null);
```

```java
// BEFORE: Poorly documented class (example)
public class Counter {
    int c;
    public Counter() { c = 0; }
    public void inc() { c++; }
    public void dec() { c--; }
    public int val() { return c; }
}

/*
Explanation (before):
- No JavaDoc, unclear field visibility implications, no thread-safety notes, method names are terse and not descriptive.
*/

Counter.main(null);
```

```java
// AFTER: Improved documentation and safer API
/**
 * Thread-safe counter with clear API and documentation.
 *
 * This class provides a simple integer counter with methods to increment,
 * decrement, and read the current value. All public methods are documented
 * with their behavior and thread-safety guarantees. The internal state is kept
 * private and access is synchronized to be safe for concurrent use.
 */
public class Counter {
    // internal counter value
    private int count;

    /**
     * Creates a new Counter initialized to zero.
     */
    public Counter() {
        this.count = 0;
    }

    /**
     * Atomically increments the counter by one.
     * @return the new value after increment
     */
    public synchronized int increment() {
        return ++count;
    }

    /**
     * Atomically decrements the counter by one.
     * @return the new value after decrement
     */
    public synchronized int decrement() {
        return --count;
    }

    /**
     * Returns the current value of the counter.
     * Note: the returned value is a snapshot; concurrent updates may change the value after return.
     * @return the current counter value
     */
    public synchronized int getValue() {
        return count;
    }
}

/*
Comparison notes (after):
- Made field private, renamed methods to descriptive names, added JavaDoc with @return tags, and documented thread-safety.
- Provided return values to aid testability and composition.
*/

Counter.main(null);
```

Hack #1: Fix the Documentation

```java
/**
 * Calculates the sum of all positive even integers in the given array.
 * <p>
 * This method loops through each element of {@code nums} and adds it to a running total
 * only if the element is greater than zero and evenly divisible by two.
 * The method then returns the resulting sum.
 * </p>
 *
 * <p><b>Preconditions:</b></p>
 * <ul>
 *   <li>{@code nums} must not be {@code null}.</li>
 *   <li>{@code nums} may contain any integers (positive, negative, or zero).</li>
 * </ul>
 *
 * <p><b>Postconditions:</b></p>
 * <ul>
 *   <li>Returns the sum of all positive even numbers in {@code nums}.</li>
 *   <li>If there are no positive even numbers, returns {@code 0}.</li>
 * </ul>
 *
 * @param nums an array of integers to evaluate (must not be null)
 * @return the sum of all positive even integers in {@code nums}; {@code 0} if none exist
 *
 * <p><b>Example usage:</b></p>
 * <pre>
 * int[] numbers = {1, 2, 3, 4, -6};
 * int result = doSomething(numbers);  // result = 6
 * </pre>
 */
public int doSomething(int[] nums) {
    if (nums == null) {
        throw new IllegalArgumentException("nums must not be null");
    }
    int result = 0;
    for (int i = 0; i < nums.length; i++) {
        if (nums[i] > 0 && nums[i] % 2 == 0) {
            result += nums[i];
        }
    }
    return result;
}
```

Popcorn Hack #2: Write Class Documentation

```java
/**
 * Manages student grades and calculates final grades based on weighted categories.
 *
 * <p>This class allows teachers to track assignments across different categories
 * (like homework, tests, projects) and calculate final grades using custom weights.</p>
 *
 * <p>Key Features:</p>
 * <ul>
 *   <li>Store assignments by category</li>
 *   <li>Apply custom weights to each category</li>
 *   <li>Track extra credit points</li>
 *   <li>Generate grade reports</li>
 * </ul>
 *
 * <p>Usage Example:</p>
 * <pre>
 * GradeBook myGrades = new GradeBook();
 * myGrades.setCategoryWeight("Homework", 0.30);
 * myGrades.setCategoryWeight("Tests", 0.70);
 * myGrades.addAssignment("Homework", "HW1", 95.0);
 * double finalGrade = myGrades.calculateFinalGrade();
 * </pre>
 *
 * @author Your Name
 * @version 1.0
 * @since 2025-10-13
 */
public class GradeBook {
    private java.util.HashMap<String, java.util.List<Double>> assignments = new java.util.HashMap<>();
    private java.util.HashMap<String, Double> categoryWeights = new java.util.HashMap<>();
    private double extraCredit = 0.0;

    /**
     * Adds an assignment score to a specific category.
     *
     * @param category the category name (e.g., "Homework", "Tests") (must not be null)
     * @param name the assignment name (not used as key in this simplified implementation)
     * @param score the score earned (0-100)
     * @throws IllegalArgumentException if category is null or score is out of range
     */
    public void addAssignment(String category, String name, double score) {
        if (category == null) throw new IllegalArgumentException("category must not be null");
        if (score < 0 || score > 100) throw new IllegalArgumentException("score must be between 0 and 100");
        assignments.computeIfAbsent(category, k -> new java.util.ArrayList<>()).add(score);
    }

    /**
     * Sets the weight for a grade category.
     *
     * @param category the category name
     * @param weight the weight as a decimal (e.g., 0.30 for 30%)
     * @throws IllegalArgumentException if category is null or weight is negative
     */
    public void setCategoryWeight(String category, double weight) {
        if (category == null) throw new IllegalArgumentException("category must not be null");
        if (weight < 0) throw new IllegalArgumentException("weight must be non-negative");
        categoryWeights.put(category, weight);
    }

    /**
     * Calculates the final weighted grade including extra credit.
     *
     * <p>Returns a percentage (0-100). If category weights do not sum to 1.0, the method
     * scales the weighted contribution according to the provided weights.</p>
     *
     * @return the final grade as a percentage
     */
    public double calculateFinalGrade() {
        double totalWeight = 0.0;
        for (double w : categoryWeights.values()) totalWeight += w;
        double score = 0.0;
        for (java.util.Map.Entry<String, Double> e : categoryWeights.entrySet()) {
            String cat = e.getKey();
            double w = e.getValue();
            java.util.List<Double> list = assignments.get(cat);
            if (list == null || list.isEmpty()) continue;
            double avg = 0.0;
            for (double s : list) avg += s;
            avg /= list.size();
            if (totalWeight > 0) score += avg * (w / totalWeight);
        }
        score += extraCredit;
        return Math.min(100.0, score);
    }

    /**
     * Generates a formatted report of all grades and categories.
     *
     * @return a String containing the grade report
     */
    public String generateReport() {
        StringBuilder sb = new StringBuilder();
        sb.append("GradeBook Report
");
        for (java.util.Map.Entry<String, java.util.List<Double>> e : assignments.entrySet()) {
            sb.append("Category: " + e.getKey() + " - Scores: " + e.getValue() + 
            );
        }
        sb.append("Final Grade: " + calculateFinalGrade() + "
");
        return sb.toString();
    }
}

GradeBook.main(null);
```

```java
/**
 * Attempts to enroll a student in a course for a given semester.
 *
 * <p>Preconditions:</p>
 * <ul>
 *   <li>{@code studentId} and {@code courseCode} must not be null or empty.</li>
 *   <li>Semester should be a positive integer representing the term.</li>
 * </ul>
 *
 * <p>Postconditions:</p>
 * <ul>
 *   <li>If the method returns {@code true}, the student has been added to the course and
 *       the enrollment transaction has been recorded.</li>
 *   <li>If the method returns {@code false}, no changes were made to student or course records.</li>
 * </ul>
 *
 * @param studentId unique identifier of the student (non-null)
 * @param courseCode code of the course to enroll in (non-null)
 * @param semester numeric semester identifier (e.g., 20251 for Spring 2025)
 * @return {@code true} if enrollment succeeded; {@code false} otherwise
 *
 * Example usage:
 * <pre>
 * boolean ok = enrollStudent("s123", "CS101", 20251);
 * </pre>
 */
public boolean enrollStudent(String studentId, String courseCode, int semester) {
    if (studentId == null || studentId.isEmpty() || courseCode == null || courseCode.isEmpty()) return false;
    if (semester <= 0) return false;

    Student student = findStudentById(studentId);
    if (student == null) return false;

    Course course = findCourseByCode(courseCode);
    if (course == null) return false;

    if (course.isFull()) return false;
    if (student.hasScheduleConflict(course)) return false;
    if (!student.hasPrerequisites(course)) return false;
    if (student.getCreditHours() + course.getCreditHours() > 18) return false;

    student.addCourse(course);
    course.addStudent(student);
    recordEnrollmentTransaction(studentId, courseCode, semester);
    return true;
}
```

Part 3: Reflection Questions

1. Why is documentation more important in team projects than solo projects?

Documentation enables shared understanding, speeds onboarding, prevents duplicated work, and clarifies design decisions. In team projects, multiple people read and modify code; clear docs reduce miscommunication and bugs.

2. Give an example of when a method SHOULD be documented and when it SHOULD NOT.

SHOULD: A method with non-obvious behavior, side effects, or complex preconditions (e.g., transaction management, concurrency control, I/O).

SHOULD NOT: Simple getters/setters or trivial one-line utility methods where the name clearly expresses the behavior.

3. Key takeaways: Documentation is communication — write for others and your future self. Use Javadoc for public APIs; include pre/postconditions for contract-like behavior. Keep docs up to date.
