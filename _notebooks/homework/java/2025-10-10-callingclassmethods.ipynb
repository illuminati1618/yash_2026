## Popcorn Hack

```java
// Popcorn Hack: Which is an instance method or static method
// Fixed code with comments explaining which lines work and which cause errors

class Bro {
    String name;
    static String catchphrase = "Sup bro?";

    public Bro(String name) {
        this.name = name;
    }

    // Instance method: belongs to the object
    public void sayHi() {
        System.out.println("Hey, I'm " + name);
    }

    // Static method: belongs to the class
    public static void sayCatchphrase() {
        System.out.println(catchphrase);
    }

    public static void main(String[] args) {
        // Will this work? YES, static method called from static context
        sayCatchphrase();

        // What about this one? ERROR: can't call instance method from static context
        // sayHi(); // <-- This line causes a compilation error

        Bro alex = new Bro("Alex");
        alex.sayHi(); // Works: instance method called on object

        // This works, but it's not recommended: static method called on instance
        alex.sayCatchphrase(); // Works, but better to use Bro.sayCatchphrase();
    }
}

// To run this class and see the output, call:
Bro.main(null);

```

# Ultimate Battle - Creature Class
This cell defines the `Fighter` class used for the Ultimate Battle homework. It includes instance variables (name, power, health), a static variable `fightDuration`, instance methods `attack()` and `printStatus()`, and class methods `strongerFighter()` and `beginBattle()`. An extra `defense` field and a `heal()` method are included as 'MAKE YOUR OWN' additions.

```java
/* Fighter class for Ultimate Battle */
class Fighter {
    String name;
    int power;
    int health;
    int defense; // extra field
    static double fightDuration = 30.0; // seconds, adjustable

    Fighter(String name, int power, int health, int defense) {
        this.name = name;
        this.power = power;
        this.health = health;
        this.defense = defense;
    }

    void attack(Fighter other) {
        int damage = Math.max(0, this.power - other.defense);
        other.health -= damage;
        System.out.println(this.name + " attacks " + other.name + " for " + damage + " damage.");
    }

    void printStatus() {
        System.out.println(name + " -- Health: " + health + ", Power: " + power + ", Defense: " + defense);
    }

    // class method to compare who is stronger based on power + health
    static int strongerFighter(Fighter a, Fighter b) {
        int scoreA = a.power + a.health;
        int scoreB = b.power + b.health;
        if (scoreA > scoreB) return 1;
        if (scoreB > scoreA) return 2;
        return 0; // tie
    }

    // class method to begin a simulated battle
    static void beginBattle(Fighter a, Fighter b) {
        System.out.println("Beginning battle between " + a.name + " and " + b.name + "! Duration limit: " + fightDuration + " seconds.");
        int rounds = 0;
        while (a.health > 0 && b.health > 0 && rounds < 100) {
            rounds++;
            // simple alternating attacks
            a.attack(b);
            if (b.health <= 0) {
                System.out.println(b.name + " has fallen!");
                break;
            }
            b.attack(a);
            if (a.health <= 0) {
                System.out.println(a.name + " has fallen!");
                break;
            }
        }
        System.out.println("Battle ended after " + rounds + " rounds.");
        a.printStatus();
        b.printStatus();
    }

    // extra utility method
    void heal(int amount) {
        this.health += amount;
        System.out.println(name + " heals for " + amount + " health.");
    }
}

Fighter f1 = new Fighter("Thor", 15, 100, 5);
f1.printStatus();
```

# Ultimate Battle - Main
This cell contains a `main` method that creates two fighters, demonstrates instance attacks and status prints, uses the static `strongerFighter()` method to compare them, prints a fact using the static `fightDuration`, and calls `beginBattle()` to simulate a full fight.

```java
public class UltimateBattle {
    public static void main(String[] args) {
        Fighter f1 = new Fighter("RoboRex", 12, 60, 3);
        Fighter f2 = new Fighter("GigaTron", 10, 70, 5);

        System.out.println("Initial status:");
        f1.printStatus();
        f2.printStatus();

        // instance attack and status
        System.out.println("\nSingle attack demo:");
        f1.attack(f2);
        f2.printStatus();

        // compare using static method
        int stronger = Fighter.strongerFighter(f1, f2);
        if (stronger == 1) System.out.println(f1.name + " is stronger based on power+health.");
        else if (stronger == 2) System.out.println(f2.name + " is stronger based on power+health.");
        else System.out.println("Both fighters are equally matched.");

        // print a fact using the static variable
        System.out.println("Fun fact: The fightDuration static variable defaults to " + Fighter.fightDuration + " seconds.");

        // start full battle
        Fighter.beginBattle(f1, f2);

        // demonstrate heal after battle (if someone survived)
        if (f1.health > 0) { f1.heal(5); f1.printStatus(); }
        if (f2.health > 0) { f2.heal(5); f2.printStatus(); }
    }
}

UltimateBattle.main(null);
```
