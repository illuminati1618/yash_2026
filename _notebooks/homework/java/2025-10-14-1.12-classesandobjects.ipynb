## Popcorn Hack (one cell)


```java
// Popcorn Hack: implement Book.printInfo and use it
class Book {
    String title;
    int pages;

    void printInfo() {
        System.out.println("Title: " + title + ", Pages: " + pages);
    }
}

class MainPopcorn {
    public static void main(String[] args) {
        Book b = new Book();
        b.title = "The Little Prince";
        b.pages = 96;
        b.printInfo();
    }
}

MainPopcorn.main(null);
```

## Homework Hack 1

```java
// Homework: define Student and create 3 instances plus an alias reference
class Student {
    String name;
    int grade;
    int pets;
    int siblings;

    Student(String name, int grade, int pets, int siblings) {
        this.name = name;
        this.grade = grade;
        this.pets = pets;
        this.siblings = siblings;
    }

    void printInfo() {
        System.out.println("Name: " + name + ", Grade: " + grade + ", Pets: " + pets + ", Siblings: " + siblings);
    }
}

class MainHomework {
    public static void main(String[] args) {
        Student s1 = new Student("Ava Green", 9, 1, 2);
        Student s2 = new Student("Liam Park", 10, 0, 1);
        Student s3 = new Student("Maya Singh", 9, 2, 0);

        // Print each student's info
        s1.printInfo();
        s2.printInfo();
        s3.printInfo();

        // Create a reference alias to s2 (nickname/reference variable)
        Student liam = s2; // liam references the same Student object as s2
        System.out.println("Reference alias output (liam):");
        liam.printInfo();
    }
}

MainHomework.main(null);
```
