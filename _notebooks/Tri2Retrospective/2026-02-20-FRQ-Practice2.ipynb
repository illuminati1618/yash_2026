## Question 1: RobotMover Class

### Problem Overview
The `RobotMover` class maintains a string representing moves made by a robot. We need to implement:
- **Part A**: A constructor that generates random moves
- **Part B**: A method to count occurrences of a substring

### Key Concepts Tested
- **Random number generation** using `Math.random()`
- **String concatenation** in loops
- **String methods**: `indexOf()`, `substring()`
- **Loop control** with string traversal

### Part A: Constructor - `RobotMover(int numMoves)`

**Task**: Initialize `moveSequence` with `numMoves` random moves, each followed by an underscore.

**Key Takeaways**:
1. **Generating random integers in a range**: `(int)(Math.random() * n)` gives values 0 to n-1
2. **Four equal choices** means we need `Math.random() * 4` to get values 0, 1, 2, or 3
3. **String building in loops**: Start with empty string, concatenate in each iteration
4. **Each move ends with underscore**: Don't forget the trailing underscore!

````python
public class RobotMover {
    private String moveSequence;

    /**
     * PART A: Constructor
     * Initializes moveSequence with numMoves moves, with an equal
     * chance of "up", "down", "left", and "right". Each move is
     * followed by an underscore ("_").
     * Precondition: numMoves > 0
     */
    public RobotMover(int numMoves) {
        // Initialize moveSequence to empty string
        moveSequence = "";
        
        // Generate numMoves random moves
        for (int i = 0; i < numMoves; i++) {
            // Generate random number 0-3 for four directions
            int direction = (int)(Math.random() * 4);
            
            // Map number to direction string
            if (direction == 0) {
                moveSequence += "up";
            } else if (direction == 1) {
                moveSequence += "down";
            } else if (direction == 2) {
                moveSequence += "left";
            } else {  // direction == 3
                moveSequence += "right";
            }
            
            // Add underscore after each move
            moveSequence += "_";
        }
    }
    
    // Getter for testing
    public String getMoveSequence() {
        return moveSequence;
    }
}

// Test Part A
System.out.println("=== Testing RobotMover Constructor (Part A) ===");
RobotMover robot1 = new RobotMover(6);
System.out.println("Robot with 6 moves: " + robot1.getMoveSequence());

RobotMover robot2 = new RobotMover(3);
System.out.println("Robot with 3 moves: " + robot2.getMoveSequence());

RobotMover robot3 = new RobotMover(10);
System.out.println("Robot with 10 moves: " + robot3.getMoveSequence());
````

### Part B: Method - `countOccurrences(String str)`

**Task**: Return the number of times `str` appears in `moveSequence`.

**Key Takeaways**:
1. **Overlapping occurrences count**: "up_up" in "up_up_up_" appears **twice** (positions 0 and 3)
2. **Use indexOf with starting position**: `indexOf(str, startIndex)` finds next occurrence after startIndex
3. **Increment by 1, not by str.length()**: This ensures we catch overlapping matches
4. **Loop until indexOf returns -1**: This signals no more occurrences found

**Algorithm**:
1. Start at index 0
2. Find `str` starting from current index
3. If found (index != -1), increment count and move start position by 1
4. Repeat until no more occurrences

````python
public class RobotMover {
    private String moveSequence;

    // Constructor from Part A
    public RobotMover(int numMoves) {
        moveSequence = "";
        for (int i = 0; i < numMoves; i++) {
            int direction = (int)(Math.random() * 4);
            if (direction == 0) {
                moveSequence += "up";
            } else if (direction == 1) {
                moveSequence += "down";
            } else if (direction == 2) {
                moveSequence += "left";
            } else {
                moveSequence += "right";
            }
            moveSequence += "_";
        }
    }
    
    // Additional constructor for testing with specific sequence
    public RobotMover(String sequence) {
        moveSequence = sequence;
    }

    /**
     * PART B: countOccurrences
     * Returns the number of times that str appears in moveSequence
     * Precondition: moveSequence is a valid sequence of moves,
     *               each followed by an underscore.
     * Postcondition: moveSequence is unchanged.
     */
    public int countOccurrences(String str) {
        int count = 0;
        int index = 0;  // Starting search position
        
        // Keep searching for str until no more found
        while (moveSequence.indexOf(str, index) != -1) {
            // Found an occurrence
            count++;
            
            // Move index forward by 1 to find overlapping occurrences
            // We use indexOf to get the actual position, then add 1
            index = moveSequence.indexOf(str, index) + 1;
        }
        
        return count;
    }
    
    public String getMoveSequence() {
        return moveSequence;
    }
}

// Test Part B with specific examples from the problem
System.out.println("=== Testing countOccurrences (Part B) ===");
RobotMover testRobot = new RobotMover("right_down_left_down_right_down_up_up_up_");
System.out.println("Testing sequence: " + testRobot.getMoveSequence());

System.out.println("\ncountOccurrences(\"right\") = " + testRobot.countOccurrences("right") + " (expected: 2)");
System.out.println("countOccurrences(\"down\") = " + testRobot.countOccurrences("down") + " (expected: 3)");
System.out.println("countOccurrences(\"right_d\") = " + testRobot.countOccurrences("right_d") + " (expected: 2)");
System.out.println("countOccurrences(\"up_up\") = " + testRobot.countOccurrences("up_up") + " (expected: 2)");
System.out.println("countOccurrences(\"left\") = " + testRobot.countOccurrences("left") + " (expected: 1)");
````

### RobotMover - Key Takeaways Summary

| Concept | Application |
|---------|-------------|
| `Math.random() * n` | Generate random int 0 to n-1 for equal probability choices |
| String concatenation in loop | Build strings incrementally (`str += ...`) |
| `indexOf(str, startIndex)` | Find substring starting from specific position |
| Overlapping matches | Move search position by 1 (not by found string length) |
| Loop until -1 | Standard pattern for finding all occurrences |

## Question 2: LapTracker Class

### Problem Overview
Design a complete class that tracks laps walked around a track with a periodic reset feature.

**Requirements**:
- Constructor takes `int` parameter for reset frequency
- Initial lap count is 0
- `addLaps(int)` method adds laps and returns total
- After every N calls to `addLaps`, the count resets to 0 before adding

### Key Concepts Tested
- **Complete class design**: Instance variables, constructor, methods
- **State management**: Tracking count AND call frequency
- **Modular arithmetic or counter**: Knowing when to reset
- **Return values**: Method returns updated count

### Design Decisions

**What instance variables do we need?**
1. `lapCount` - Current total laps
2. `resetFrequency` - How often to reset (from constructor)
3. `callCount` - Number of times `addLaps` has been called

**Logic for addLaps**:
1. Increment call counter
2. Check if it's time to reset (callCount reaches resetFrequency)
3. If reset needed: set lapCount to 0 and reset callCount
4. Add the new laps
5. Return the new total

````python
/**
 * LapTracker class - Complete implementation
 * Tracks laps with periodic reset functionality
 */
public class LapTracker {
    // Instance variables
    private int lapCount;       // Current total lap count
    private int resetFrequency; // Reset after this many calls to addLaps
    private int callCount;      // Number of times addLaps has been called since last reset
    
    /**
     * Constructor
     * Creates a LapTracker that resets after resetAfter calls to addLaps
     * Precondition: resetAfter > 0
     */
    public LapTracker(int resetAfter) {
        this.lapCount = 0;              // Start with 0 laps
        this.resetFrequency = resetAfter;
        this.callCount = 0;             // No calls yet
    }
    
    /**
     * Adds laps to the total and returns the new total
     * Resets lap count to 0 after every resetFrequency calls
     * Precondition: numLaps > 0
     * @param numLaps - number of laps to add
     * @return the updated lap count
     */
    public int addLaps(int numLaps) {
        // Increment the call counter
        callCount++;
        
        // Check if it's time to reset
        if (callCount > resetFrequency) {
            // Reset the lap count and call counter
            lapCount = 0;
            callCount = 1;  // This is the first call of the new cycle
        }
        
        // Add the new laps
        lapCount += numLaps;
        
        // Return the updated count
        return lapCount;
    }
}

// Test with examples from the problem
System.out.println("=== Testing LapTracker ===");
int count;

// Test amy (resets every 3 calls)
LapTracker amy = new LapTracker(3);
System.out.println("\nAmy's tracker (resets every 3 calls):");

count = amy.addLaps(8);
System.out.println("addLaps(8) = " + count + " (expected: 8)");

count = amy.addLaps(12);
System.out.println("addLaps(12) = " + count + " (expected: 20)");

count = amy.addLaps(10);
System.out.println("addLaps(10) = " + count + " (expected: 30) - Next call will reset");

count = amy.addLaps(11);
System.out.println("addLaps(11) = " + count + " (expected: 11) - After reset");

count = amy.addLaps(5);
System.out.println("addLaps(5) = " + count + " (expected: 16)");

count = amy.addLaps(13);
System.out.println("addLaps(13) = " + count + " (expected: 29) - Next call will reset");

count = amy.addLaps(7);
System.out.println("addLaps(7) = " + count + " (expected: 7) - After reset");
````

````python
// Continue testing with eve (resets every 2 calls)
public class LapTracker {
    private int lapCount;
    private int resetFrequency;
    private int callCount;
    
    public LapTracker(int resetAfter) {
        this.lapCount = 0;
        this.resetFrequency = resetAfter;
        this.callCount = 0;
    }
    
    public int addLaps(int numLaps) {
        callCount++;
        if (callCount > resetFrequency) {
            lapCount = 0;
            callCount = 1;
        }
        lapCount += numLaps;
        return lapCount;
    }
}

System.out.println("\n=== Testing Eve's tracker (resets every 2 calls) ===");
LapTracker eve = new LapTracker(2);

int count = eve.addLaps(20);
System.out.println("addLaps(20) = " + count + " (expected: 20)");

count = eve.addLaps(13);
System.out.println("addLaps(13) = " + count + " (expected: 33) - Next call will reset");

count = eve.addLaps(10);
System.out.println("addLaps(10) = " + count + " (expected: 10) - After reset");
````

### LapTracker - Key Takeaways Summary

| Concept | Application |
|---------|-------------|
| **State tracking** | Multiple instance variables work together to maintain complex state |
| **Counter-based reset** | Use a counter that gets reset along with the main value |
| **Timing of operations** | Order matters: check reset BEFORE adding, update counter BEFORE checking |
| **Complete class design** | Instance variables, constructor, methods with proper access |

**Common Mistakes to Avoid**:
- Forgetting to reset the call counter when lap count resets
- Checking `callCount == resetFrequency` instead of `> resetFrequency`
- Adding laps before checking for reset

## Question 3: PlayerAnalysis - playerWithClosestScore

### Problem Overview
Find the player whose score is closest to a target score from an ArrayList of Player objects.

### Key Concepts Tested
- **ArrayList traversal**
- **Finding minimum/closest value**
- **Absolute value** for distance calculation
- **Tracking both value AND associated data** (distance and ID)

### Algorithm Design

**Key Insight**: "Closest" means minimum absolute difference between score and target.

**Algorithm**:
1. Initialize with first player as "best so far"
2. Calculate distance: `Math.abs(score - targetScore)`
3. Loop through all players
4. If current player is closer, update best player
5. Return the best player's ID

**Why start with first player?**
- We're guaranteed list is not empty
- Avoids using arbitrary initial values like `Integer.MAX_VALUE` (though that works too)

````python
import java.util.ArrayList;

// Player class (as given)
class Player {
    private String id;
    private int score;
    
    public Player(String id, int score) {
        this.id = id;
        this.score = score;
    }
    
    public String getID() {
        return id;
    }
    
    public int getScore() {
        return score;
    }
}

// PlayerAnalysis class with the method to implement
class PlayerAnalysis {
    private ArrayList<Player> playerList;
    
    // Constructor for testing
    public PlayerAnalysis(ArrayList<Player> players) {
        this.playerList = players;
    }
    
    /**
     * Returns the ID of the player whose score is closest to targetScore
     * Preconditions: playerList is not null, not empty, no nulls
     * Postcondition: playerList is unchanged
     */
    public String playerWithClosestScore(int targetScore) {
        // Start with the first player as the closest
        Player closestPlayer = playerList.get(0);
        int closestDistance = Math.abs(closestPlayer.getScore() - targetScore);
        
        // Check all other players
        for (int i = 1; i < playerList.size(); i++) {
            Player current = playerList.get(i);
            int currentDistance = Math.abs(current.getScore() - targetScore);
            
            // If this player is closer, update our best
            if (currentDistance < closestDistance) {
                closestPlayer = current;
                closestDistance = currentDistance;
            }
        }
        
        return closestPlayer.getID();
    }
}

// Set up test data from the problem
ArrayList<Player> players = new ArrayList<>();
players.add(new Player("DJK", 1090));
players.add(new Player("CJL", 2800));
players.add(new Player("JOY", 2000));
players.add(new Player("BEN", 500));
players.add(new Player("PTT", 3500));
players.add(new Player("JAY", 4500));

PlayerAnalysis analysis = new PlayerAnalysis(players);

// Test with examples from the problem
System.out.println("=== Testing playerWithClosestScore ===");
System.out.println("Player list: DJK(1090), CJL(2800), JOY(2000), BEN(500), PTT(3500), JAY(4500)");
System.out.println();

System.out.println("playerWithClosestScore(1000) = \"" + analysis.playerWithClosestScore(1000) + "\" (expected: DJK)");
System.out.println("playerWithClosestScore(2000) = \"" + analysis.playerWithClosestScore(2000) + "\" (expected: JOY)");
System.out.println("playerWithClosestScore(3000) = \"" + analysis.playerWithClosestScore(3000) + "\" (expected: CJL)");
System.out.println("playerWithClosestScore(4000) = \"" + analysis.playerWithClosestScore(4000) + "\" (expected: PTT or JAY)");
System.out.println("playerWithClosestScore(5000) = \"" + analysis.playerWithClosestScore(5000) + "\" (expected: JAY)");
````

### Alternative Implementation: Using Enhanced For Loop

````python
import java.util.ArrayList;

class Player {
    private String id;
    private int score;
    
    public Player(String id, int score) {
        this.id = id;
        this.score = score;
    }
    
    public String getID() { return id; }
    public int getScore() { return score; }
}

class PlayerAnalysis {
    private ArrayList<Player> playerList;
    
    public PlayerAnalysis(ArrayList<Player> players) {
        this.playerList = players;
    }
    
    /**
     * Alternative implementation using enhanced for loop
     * and Integer.MAX_VALUE initialization
     */
    public String playerWithClosestScore(int targetScore) {
        String closestID = "";
        int closestDistance = Integer.MAX_VALUE;  // Start with largest possible distance
        
        // Enhanced for loop - cleaner syntax
        for (Player player : playerList) {
            int distance = Math.abs(player.getScore() - targetScore);
            
            if (distance < closestDistance) {
                closestID = player.getID();
                closestDistance = distance;
            }
        }
        
        return closestID;
    }
}

// Quick test
ArrayList<Player> players = new ArrayList<>();
players.add(new Player("DJK", 1090));
players.add(new Player("CJL", 2800));
players.add(new Player("JOY", 2000));

PlayerAnalysis analysis = new PlayerAnalysis(players);
System.out.println("Using enhanced for loop:");
System.out.println("playerWithClosestScore(2000) = \"" + analysis.playerWithClosestScore(2000) + "\"");
````

### PlayerAnalysis - Key Takeaways Summary

| Concept | Application |
|---------|-------------|
| **Math.abs()** | Calculate distance regardless of direction (above or below target) |
| **Track two things** | Need to track both the minimum distance AND the corresponding ID |
| **Initialization strategies** | Either use first element or use `Integer.MAX_VALUE` |
| **Enhanced for loop** | Cleaner when you don't need the index |

**Common Mistakes to Avoid**:
- Forgetting `Math.abs()` - would fail if target is between two scores
- Returning the score instead of the ID
- Not handling ties (problem says any ID is acceptable)

## Question 4: WordGrid - countOrderedRows

### Problem Overview
Count rows in a 2D String array where strings are ordered by length (shortest to longest).

### Key Concepts Tested
- **2D Array traversal** (row by row, then column by column)
- **Comparing adjacent elements** within a row
- **String length comparison**
- **Boolean flag pattern** for checking row validity

### Algorithm Design

**Definition of "ordered"**: Each string's length >= previous string's length

**Algorithm**:
1. Initialize counter for ordered rows
2. For each row:
   - Assume the row is ordered (flag = true)
   - Compare each element with the previous element
   - If any element is shorter than the previous, row is NOT ordered (flag = false)
   - If row passed all checks, increment counter
3. Return the counter

**Key Insight**: We check from column 1 to end, comparing with column-1

````python
public class WordGrid {
    private String[][] grid;
    
    // Constructor for testing
    public WordGrid(String[][] grid) {
        this.grid = grid;
    }
    
    /**
     * Returns the number of rows in grid that are ordered from
     * shortest string to longest string
     * Preconditions: grid has at least one row and at least two columns.
     *                No elements in grid are null.
     * Postcondition: grid is unchanged.
     */
    public int countOrderedRows() {
        int count = 0;  // Count of ordered rows
        
        // Iterate through each row
        for (int row = 0; row < grid.length; row++) {
            boolean isOrdered = true;  // Assume this row is ordered
            
            // Check each pair of adjacent elements in the row
            for (int col = 1; col < grid[row].length; col++) {
                // Compare current element's length with previous element's length
                if (grid[row][col].length() < grid[row][col - 1].length()) {
                    // Current is shorter than previous - NOT ordered
                    isOrdered = false;
                    break;  // No need to check rest of row
                }
            }
            
            // If row passed all checks, it's ordered
            if (isOrdered) {
                count++;
            }
        }
        
        return count;
    }
}

// Test with the example from the problem
String[][] testGrid = {
    {"cat", "crane", "people", "hamster"},    // Row 0: 3, 5, 6, 7 - ORDERED
    {"dog", "slate", "bat", "green"},         // Row 1: 3, 5, 3, 5 - NOT ordered (5 > 3)
    {"blue", "audio", "adieu", "snazzy"},     // Row 2: 4, 5, 5, 6 - ORDERED
    {"yellow", "shirt", "hat", "store"},      // Row 3: 6, 5, 3, 5 - NOT ordered (6 > 5)
    {"about", "media", "stone", "canoe"}      // Row 4: 5, 5, 5, 5 - ORDERED (equal counts)
};

WordGrid wordGrid = new WordGrid(testGrid);

System.out.println("=== Testing countOrderedRows ===");
System.out.println("Grid contents:");
for (int i = 0; i < testGrid.length; i++) {
    System.out.print("Row " + i + ": ");
    for (String s : testGrid[i]) {
        System.out.print(s + "(" + s.length() + ") ");
    }
    System.out.println();
}

int result = wordGrid.countOrderedRows();
System.out.println("\ncountOrderedRows() = " + result + " (expected: 3)");
System.out.println("Ordered rows are: 0, 2, and 4");
````

### Visual Breakdown of the Example

| Row | Elements | Lengths | Ordered? |
|-----|----------|---------|----------|
| 0 | cat → crane → people → hamster | 3 → 5 → 6 → 7 |  Yes (increasing) |
| 1 | dog → slate → bat → green | 3 → 5 → **3** → 5 |  No (5 > 3) |
| 2 | blue → audio → adieu → snazzy | 4 → 5 → 5 → 6 |  Yes (non-decreasing) |
| 3 | yellow → shirt → hat → store | **6** → 5 → 3 → 5 |  No (6 > 5) |
| 4 | about → media → stone → canoe | 5 → 5 → 5 → 5 |  Yes (equal is allowed) |

**Result: 3 ordered rows**

````python
// Alternative implementation without boolean flag (using method extraction)
public class WordGrid {
    private String[][] grid;
    
    public WordGrid(String[][] grid) {
        this.grid = grid;
    }
    
    /**
     * Helper method: checks if a single row is ordered
     */
    private boolean isRowOrdered(int row) {
        for (int col = 1; col < grid[row].length; col++) {
            if (grid[row][col].length() < grid[row][col - 1].length()) {
                return false;
            }
        }
        return true;
    }
    
    /**
     * Alternative implementation using helper method
     */
    public int countOrderedRows() {
        int count = 0;
        
        for (int row = 0; row < grid.length; row++) {
            if (isRowOrdered(row)) {
                count++;
            }
        }
        
        return count;
    }
}

// Quick test
String[][] grid2 = {
    {"a", "bb", "ccc"},      // Ordered: 1 -> 2 -> 3
    {"xxx", "yy", "z"},      // Not ordered: 3 -> 2 -> 1
    {"hi", "hi", "hi"}       // Ordered: 2 -> 2 -> 2 (equal allowed)
};

WordGrid wg2 = new WordGrid(grid2);
System.out.println("\n=== Additional Test ===");
System.out.println("countOrderedRows() = " + wg2.countOrderedRows() + " (expected: 2)");
````

### WordGrid - Key Takeaways Summary

| Concept | Application |
|---------|-------------|
| **2D array traversal** | Outer loop for rows, inner loop for columns |
| **Adjacent element comparison** | Start inner loop at 1, compare `[col]` with `[col-1]` |
| **Boolean flag pattern** | Assume true, set false if violation found |
| **Early exit with break** | Once row fails, no need to check remaining elements |
| **>= vs >** | "Ordered" means non-decreasing (equal lengths are OK) |

**2D Array Access Pattern**:
```java
grid.length          // Number of rows
grid[row].length     // Number of columns in that row
grid[row][col]       // Element at position (row, col)
```
